#### ***Part B: Support Vector Machines for 10-class classification***

```python
import time
import dataProccessing as dp
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

**CIFAR-10 dataset preparation**

Dataset was maintained to 10 classes:

```python
# 1. Load CIFAR-10 dataset
x_train_raw, y_train_raw, x_test_raw, y_test_raw = dp.load_cifar10()

# 2. Load CIFAR-10 original class names
class_names = dp.get_class_names()

# 3. Split training data into training and validation sets
x_train_split, y_train_split, x_val_split, y_val_split = dp.split_train_validation(x_train_raw, y_train_raw, val_size=0.4)

# 4. Flatten and normalize CIFAR-10 images
x_train_fl, x_val_fl, x_test_fl = dp.flatten_and_normalize(x_train_split, x_val_split, x_test_raw)

# 5. Apply PCA for dimensionality reduction
x_train, x_val, x_test = dp.apply_pca(x_train_fl, x_val_fl, x_test_fl, n_components=0.9)

# 6. Rename variables for clarity
y_train = y_train_split
y_val = y_val_split
y_test = y_test_raw
```

##### ***Linear Kernel***
Testing Linear Kernel SVM for C=1.0:

```python
# Experiment 1: Linear Kernel SVM with C=1.0
print("\nRunning Experiment 1: Linear Kernel SVM for C=1.0")
print("="*70)

start_time = time.time()

# Train Linear SVM
svm_linear = SVC(kernel='linear', C=1.0, random_state=42)
svm_linear.fit(x_train, y_train)
training_time = time.time() - start_time
print(f"\nTraining time: {training_time:.2f} seconds")

# Evaluate Linear SVM
y_train_pred = svm_linear.predict(x_train)
train_accuracy = accuracy_score(y_train, y_train_pred)
print(f"Training Accuracy: {train_accuracy:.2%}")

y_val_pred = svm_linear.predict(x_val)
val_accuracy = accuracy_score(y_val, y_val_pred)
print(f"Validation Accuracy: {val_accuracy:.2%}")

print("\n" + "="*70)
```

##### ***Polynomial Kernel***
Testing Polynomial Kernel SVM for C=1.0 and Degree=2:

```python
# Experiment 2: Polynomial Kernel with C=1.0 and Degree=2
print(f"\nRunning Experiment 2: Polynomial Kernel SVM for C=1.0 and Degree=2")
print("="*70)

start_time = time.time()
    
# Train Polynomial SVM
svm_poly = SVC(kernel='poly', degree=2, C=1.0, coef0=1, random_state=42)
svm_poly.fit(x_train, y_train)
training_time = time.time() - start_time
print(f"\nTraining time: {training_time:.2f} seconds")
    
# Evaluate Polynomial SVM on training set
y_train_pred = svm_poly.predict(x_train)
train_accuracy = accuracy_score(y_train, y_train_pred)
print(f"Training Accuracy: {train_accuracy:.2%}")

# Evaluate Polynomial SVM on validation set
y_val_pred = svm_poly.predict(x_val)
val_accuracy = accuracy_score(y_val, y_val_pred)
print(f"Validation Accuracy: {val_accuracy:.2%}")

print("\n" + "="*70)
```

##### ***RBF Kernel***
Testing RBF Kernel SVM for C=1.0 and gamma=0.01:

```python
# Experiment 3: RBF Kernel with C=1.0 and gamma=0.01
print("\nRunning Experiment 3: RBF Kernel SVM for C=1.0 and gamma=0.01")
print("="*70)

start_time = time.time()

# Train RBF SVM
svm_rbf = SVC(kernel='rbf', C=1.0, gamma=0.01, random_state=42, decision_function_shape='ovr')
svm_rbf.fit(x_train, y_train)
training_time = time.time() - start_time
print(f"\nTraining time: {training_time:.2f} seconds")

# Evaluate RBF SVM on training set
y_train_pred = svm_rbf.predict(x_train)
train_accuracy = accuracy_score(y_train, y_train_pred)
print(f"Training Accuracy: {train_accuracy:.2%}")

# Evaluate RBF SVM on validation set
y_val_pred = svm_rbf.predict(x_val)
val_accuracy = accuracy_score(y_val, y_val_pred)
print(f"Validation Accuracy: {val_accuracy:.2%}")

print("\n" + "="*70)
```

##### ***Sigmoid Kernel***
Testing Sigmoid Kernel SVM for C=1.0 and gamma='scale':

```python
# Experiment 4: Sigmoid Kernel SVM with C=1.0 and gamma='scale'
print("\nRunning Experiment 4: Sigmoid Kernel SVM for C=1.0 and gamma='scale'")
print("="*70)

start_time = time.time()

# Train Sigmoid SVM for gamma='scale'
svm_sigmoid = SVC(kernel='sigmoid', C=1.0, gamma='scale', coef0=0, random_state=42, decision_function_shape='ovr')
svm_sigmoid.fit(x_train, y_train)
training_time = time.time() - start_time
print(f"\nTraining time: {training_time:.2f} seconds")

# Evaluate Sigmoid SVM for gamma='scale'
y_train_pred = svm_sigmoid.predict(x_train)
train_accuracy = accuracy_score(y_train, y_train_pred)
print(f"Training Accuracy: {train_accuracy:.2%}")

y_val_pred = svm_sigmoid.predict(x_val)
val_accuracy = accuracy_score(y_val, y_val_pred)
print(f"Validation Accuracy: {val_accuracy:.2%}")

print("\n" + "="*70)
```
